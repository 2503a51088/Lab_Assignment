```python
import pandas as pd
import plotly.express as px

# Load your dataset
df = pd.read_csv("Technology_Trends_Dashboard.csv")

# Clean up: fill missing values
df['category'] = df['category'].fillna('none')
df['job_postings'] = df['job_postings'].fillna(0)
df['mentions'] = df['mentions'].fillna(0)
df['average_salary_usd'] = df['average_salary_usd'].fillna(0)
```

```python
fig = px.histogram(df, x="popularity_score")
fig.update_layout(xaxis_rangeslider_visible=True)
fig.show()
```

```python
fig = px.scatter(
    df, x="popularity_score", y="demand_index",
    color="category",
    hover_data=["technology","year","region"]
)
fig.update_yaxes(type="log")  # try log scale for skew
fig.show()
```

```python
fig = px.box(
    df, x="category", y="popularity_score",
    points="all"
)
fig.show()
```

```python
fig = px.scatter_3d(
    df, x="popularity_score", y="demand_index", z="average_salary_usd",
    color="category",
    hover_data=["technology","year","region"]
)
fig.show()
```

```python
import pandas as pd
import plotly.express as px

# Load your dataset
df = pd.read_csv("Technology_Trends_Dashboard.csv")

# Columns that should be numeric
numeric_cols = ["popularity_score", "demand_index", "job_postings", "mentions", "average_salary_usd"]

# Convert to numeric, coerce errors to NaN
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Replace NaN with 0 for plotting
df[numeric_cols] = df[numeric_cols].fillna(0)
```

```python
df['job_postings'] = df['job_postings'].clip(lower=0)
fig = px.scatter(
    df,
    x="popularity_score",
    y="demand_index",
    animation_frame="year",   # yearly frames
    size="job_postings",      # bubble size
    color="category",
    hover_data=["technology","region","experience_level"]
)

fig.update_layout(
    xaxis=dict(range=[0, 120]),
    yaxis=dict(range=[0, 120])
)

fig.show()
```

```python
fig = px.scatter(
    df,
    x="popularity_score",
    y="demand_index",   # default y-axis
    color="category",
    hover_data=["technology","region","year"]
)

fig.update_layout(
    updatemenus=[
        dict(
            buttons=list([
                dict(label="Demand Index", method="update",
                     args=[{"y":[df["demand_index"]]}, {"yaxis":{"title":"Demand Index"}}]),
                dict(label="Average Salary", method="update",
                     args=[{"y":[df["average_salary_usd"]]}, {"yaxis":{"title":"Average Salary"}}]),
                dict(label="Mentions", method="update",
                     args=[{"y":[df["mentions"]]}, {"yaxis":{"title":"Mentions"}}])
            ]),
            direction="down",
            showactive=True
        )
    ]
)

fig.show()
```
